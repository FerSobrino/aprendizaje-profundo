## Tarea 5: Regularización — material de práctica (no calificado)

Esta tarea **no se entrega ni se califica**. Es material de práctica y preparación: los exámenes evalúan la teoría detrás de estos experimentos (temarios en `Examenes/`) y el proyecto de colaboración con IA usa este mismo protocolo experimental.

Sugerencias para sacarle provecho:
1. Puedes usar LLMs o agentes de IA libremente. Pero recuerda: los exámenes son en clase y a mano, evalúan que *tú* entiendas.
2. Antes de correr cada experimento, escribe qué esperas observar y por qué; después compara con lo que obtuviste.
3. Asegúrate de entender cada línea del código (tuyo o del agente); consulta la documentación de PyTorch.
4. Mantén el protocolo experimental del curso: semilla fija, mismos splits, un factor variado a la vez, curvas + tabla resumen + interpretación.

1. Weight decay (L2) y optimizadores. Acá abajo el setup para este problema: 
     - Dataset: MNIST (clasificación) o un problema de regresión pequeño si prefieres (usar MNIST facilita comparar).
     - Red neuronal superficial entrada 28×28 plano → hidden 512 → output 10 sin BatchNormalization 
     - Hiperparámetros base: epochs=30, batch_size=128, lr=1e-2 (ajustar por optimizador), seed=1234.

Experimentos: 
- Para cada optimizador en {SGD, SGD+momentum(0.9), Adam} entrena 3 runs con weight_decay ∈ {0, 1e-4, 1e-3} (mantén lr fijo por optimizador).
- Registra por época: train_loss, val_loss, train_acc, val_acc, y param_norm = ||θ||_2.

¿Qué espero ver?
- Curvas train/val loss y acc (por optimizador, una figura por optimizador mostrando las tres curvas de weight_decay).
- tabla final con val_acc, param_norm y ECE para cada combinación
- Explicación ¿cómo cambia la trayectoria de optimización con weight decay? ¿Interacciona con el optimizador (p. ej. Adam vs SGD)?

2. L1 y sparsity / pruning. Acá abajo el setup para este problema:
    - Dataset: problema de regresión sintética (p. ej. make_reg_synthetic con input_dim=20) o MNIST (si quieres clasificación).
    - Red neuronal profunda con dos capas intermedias, salida depende de que dataset escojas 
    - Hiperparámetros: epochs=30, batch_size=64, lr=1e-2. 

Experimentos:
- Entrena modelos con L1 penalty manual λ * sum(|w|) para λ ∈ {0, 1e-5, 1e-4, 1e-3}. Implementa L1 sumando al loss.
- Entrena un baseline (no L1). Luego aplica pruning estructurado o unstructured: torch.nn.utils.prune.l1_unstructured(module, name='weight', amount=p) para p∈{0.3,0.5,0.8} y evalúa inmediatamente y después de fine-tuning (p.ej. 10 epochs con lr 1e-3)
- Mide: RMSE (regresión) o accuracy (clasificación), porcentaje de ceros (sparsity) por capa, histograma de pesos (epoch 0, final, post-prune).

¿Qué espero ver?
- Curvas train/val loss para diferentes λ y para baseline+prune.
- Tabla: λ / %prune → final RMSE/accuracy y sparsity.
- ¿qué trade-off observas entre sparsity y desempeño? ¿Prune+fine-tune recupera desempeño?

3. Data Augmentation en régimen de pocos datos. Acá abajo el setup para este problema:
    - Dataset: MNIST reducido — usa sólo N_train = 2000 ejemplos.
    - Red neuronal superficial, la misma que en el ejercicio 1 
    - Hiperparámetros: epochs=40, batch_size=64, lr=1e-2.

Experimentos:
- Entrena sin augmentation (baseline).
- Entrena con augmentations conservadores para dígitos: small rotations (±10°), random translation (±2 pixels), random affine scale (0.95–1.05).
- Entrena con Gaussian input-noise (agregar ruido gaussiano a pixels) como alternativa.
- Mide val_acc, curvas de loss y param_norm

¿Qué espero ver?
- Curvas comparativas (baseline vs augmentation vs noise).
- Tabla resumen: val_acc, train_acc, gap = train_acc - val_acc.
- ¿qué técnica mejor compensó pocos datos y por qué?
